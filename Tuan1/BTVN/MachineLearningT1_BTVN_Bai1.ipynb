```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Tao cung mot bo dien tich cho tat ca thi nghiem
rng = np.random.default_rng(42)
X = rng.uniform(30, 200, (300, 1))
nhieu_goc = rng.normal(0, 1, 300)

# Chia chi so co dinh de so sanh cong bang
idx_train, idx_test = train_test_split(
    np.arange(300), test_size=0.2, random_state=42
)

ket_qua = []
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Thu nghiem 4 muc do nhieu
for ax, noise in zip(axes.ravel(), [0, 100, 300, 600]):
    y = 500 + 20 * X.ravel() + noise * nhieu_goc

    model = LinearRegression()
    model.fit(X[idx_train], y[idx_train])
    y_pred = model.predict(X[idx_test])

    mse = mean_squared_error(y[idx_test], y_pred)
    r2 = r2_score(y[idx_test], y_pred)

    ket_qua.append({
        "Do nhieu": noise,
        "MSE": round(mse, 2),
        "R2": round(r2, 4)
    })

    ax.scatter(X, y, alpha=0.35)
    x_line = np.linspace(30, 200, 100).reshape(-1, 1)
    ax.plot(x_line, model.predict(x_line), color="red")
    ax.set_title(f"Do nhieu = {noise}, R2 = {r2:.3f}")
    ax.set_xlabel("Dien tich (m2)")
    ax.set_ylabel("Gia nha (trieu VND)")

plt.tight_layout()
plt.show()

bang = pd.DataFrame(ket_qua)
print("BANG SO SANH:")
print(bang.to_string(index=False))
```
